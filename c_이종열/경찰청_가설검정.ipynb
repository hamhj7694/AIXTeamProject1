## 0-4. 원본 데이터 기본 구조 확인

값을 변경하기 전에 `head`, `shape`, 컬럼명, `info`, 기초 통계를 확인합니다. 반복 출력만 줄이기 위해 작은 확인 함수를 사용합니다.

```python
def show_basic_structure(name, df):
    print('\n' + '=' * 80)
    print(name)
    print('shape:', df.shape)
    print('columns:', df.columns.tolist())
    display(df.head())
    print('\n[info]')
    df.info()
    print('\n[숫자형 기초 통계]')
    display(df.describe())
    print('\n[전체 컬럼 요약]')
    display(df.describe(include='all').transpose())

for name, df in raw_dataframes.items():
    show_basic_structure(name, df)
```

## 0-5. 결측치 확인

컬럼별 결측 개수와 결측률을 함께 확인합니다. 이 단계에서는 결측 행을 삭제하거나 채우지 않습니다.

```python
for name, df in raw_dataframes.items():
    missing_report = pd.DataFrame({
        '결측 개수': df.isna().sum(),
        '결측률(%)': (df.isna().mean() * 100).round(2),
    })
    print('\n', name)
    display(missing_report)
```

## 0-6. 중복 확인

## 0-7. 연도 및 지역 값 확인

경찰청 현황·연령별 자료의 연도 범위와 두 지역 자료의 시도청 범주가 일치하는지 확인합니다.

```python
print('[경찰청 연도 범위]')
print('보이스피싱 현황:', sorted(df_police_status_raw['구분'].unique()))
print('연령별 현황:', sorted(df_police_age_raw['구분'].unique()))

print('\n[경찰청 지역 범주 일치 여부]')
region_count_set = set(df_police_region_count_raw['시도청'])
region_amount_set = set(df_police_region_amount_raw['시도청'])
print('발생 현황에만 존재:', sorted(region_count_set - region_amount_set))
print('피해금액 현황에만 존재:', sorted(region_amount_set - region_count_set))
```

## 0-8. dtype 및 값 형식 확인

컬럼명·문자열 앞뒤 공백, 숫자형이어야 할 값의 변환 실패 여부, 연도와 피해금액 단위를 확인합니다. 경찰청 보이스피싱 현황의 피해액 컬럼은 이름과 원문 기준 억원 단위입니다.

```python
for name, df in raw_dataframes.items():
    column_space_count = sum(column != column.strip() for column in df.columns)
    string_space_rows = {}
    for column in df.select_dtypes(include=['object', 'string']).columns:
        values = df[column].dropna().astype(str)
        count = int(values.ne(values.str.strip()).sum())
        if count > 0:
            string_space_rows[column] = count
    print(f'{name}: 컬럼명 공백={column_space_count}, 문자열 값 공백={string_space_rows}')
```

```python
expected_numeric_columns = {
    '경찰청 보이스피싱 현황': df_police_status_raw.columns.tolist(),
    '경찰청 피해자 연령별 현황': df_police_age_raw.columns.tolist(),
    '경찰청 지역별 발생 현황': df_police_region_count_raw.columns[1:].tolist(),
    '경찰청 지역별 피해금액': df_police_region_amount_raw.columns[1:].tolist(),
}

for name, columns in expected_numeric_columns.items():
    df = raw_dataframes[name]
    print(f'\n[{name}]')
    for column in columns:
        converted = pd.to_numeric(df[column], errors='coerce')
        new_missing = int((converted.isna() & df[column].notna()).sum())
        print(f'{column}: 현재 dtype={df[column].dtype}, 숫자 변환 실패={new_missing}건')
```

```python
print('[피해금액 단위 확인]')
print('경찰청 보이스피싱 현황 피해액 단위: 컬럼명 기준 억원')
print('경찰청 지역별 피해금액 단위: 원문 문서에서 별도 명시가 없어 사람이 확인 필요')
```

## 0-9. 최소 전처리

원본 DataFrame을 보호하기 위해 복사본에서만 작업합니다. 컬럼명과 문자열의 앞뒤 공백을 안전하게 제거하고, 숫자형이어야 하는 컬럼만 명시적으로 변환합니다. 의미가 다른 범주를 임의로 합치거나 행을 삭제하지 않습니다.

```python
df_police_status = df_police_status_raw.copy()
df_police_age = df_police_age_raw.copy()
df_police_region_count = df_police_region_count_raw.copy()
df_police_region_amount = df_police_region_amount_raw.copy()

prepared_dataframes = {
    '경찰청 보이스피싱 현황': df_police_status,
    '경찰청 피해자 연령별 현황': df_police_age,
    '경찰청 지역별 발생 현황': df_police_region_count,
    '경찰청 지역별 피해금액': df_police_region_amount,
}

for df in prepared_dataframes.values():
    df.columns = df.columns.str.strip()
    for column in df.select_dtypes(include=['object', 'string']).columns:
        df[column] = df[column].str.strip()
```

```python
# 경찰청: 연도를 뜻하는 '구분'을 분석 목적이 분명한 '연도'로 변경합니다.
df_police_status = df_police_status.rename(columns={'구분': '연도'})
df_police_age = df_police_age.rename(columns={'구분': '연도'})

police_status_numeric = df_police_status.columns.tolist()
police_age_numeric = df_police_age.columns.tolist()
police_region_count_numeric = df_police_region_count.columns[1:].tolist()
police_region_amount_numeric = df_police_region_amount.columns[1:].tolist()

for column in police_status_numeric:
    df_police_status[column] = pd.to_numeric(df_police_status[column], errors='coerce')
for column in police_age_numeric:
    df_police_age[column] = pd.to_numeric(df_police_age[column], errors='coerce')
for column in police_region_count_numeric:
    df_police_region_count[column] = pd.to_numeric(df_police_region_count[column], errors='coerce')
for column in police_region_amount_numeric:
    df_police_region_amount[column] = pd.to_numeric(df_police_region_amount[column], errors='coerce')

print('숫자형 변환 후 새 결측 확인')
for name, df in {
    '경찰청 보이스피싱 현황': df_police_status,
    '경찰청 피해자 연령별 현황': df_police_age,
    '경찰청 지역별 발생 현황': df_police_region_count,
    '경찰청 지역별 피해금액': df_police_region_amount,
}.items():
    print(name, int(df.isna().sum().sum()))
```

## 0-10. 필요한 파생변수 확인 및 생성

최신 경찰청 가설 문서의 P1·P2에 필요한 `전체_발생건수`, `전체_피해액_억원`만 생성합니다. 지역별 건당 피해금액은 분석 단계의 계산이므로 아직 만들지 않습니다.

```python
if '전체_발생건수' not in df_police_status.columns:
    df_police_status['전체_발생건수'] = (
        df_police_status['기관사칭형_발생건수']
        + df_police_status['대출사기형_발생건수']
    )

if '전체_피해액_억원' not in df_police_status.columns:
    df_police_status['전체_피해액_억원'] = (
        df_police_status['기관사칭형_피해액_억원']
        + df_police_status['대출사기형_피해액_억원']
    )

display(df_police_status[['연도', '전체_발생건수', '전체_피해액_억원']])
```

## 0-11. 전처리 결과 최종 확인

전처리 전후 shape, 결측, 완전중복, dtype과 생성 변수를 비교합니다.

```python
final_dataframes = {
    '경찰청 보이스피싱 현황': df_police_status,
    '경찰청 피해자 연령별 현황': df_police_age,
    '경찰청 지역별 발생 현황': df_police_region_count,
    '경찰청 지역별 피해금액': df_police_region_amount,
}

comparison_rows = []
raw_for_comparison = {
    '경찰청 보이스피싱 현황': df_police_status_raw,
    '경찰청 피해자 연령별 현황': df_police_age_raw,
    '경찰청 지역별 발생 현황': df_police_region_count_raw,
    '경찰청 지역별 피해금액': df_police_region_amount_raw,
}

for name, final_df in final_dataframes.items():
    raw_df = raw_for_comparison[name]
    comparison_rows.append({
        '데이터': name,
        '전처리 전 shape': str(raw_df.shape),
        '전처리 후 shape': str(final_df.shape),
        '최종 결측': int(final_df.isna().sum().sum()),
        '최종 완전중복 추가 행': int(final_df.duplicated().sum()),
        '원본에서 삭제한 행': 0,
    })

display(pd.DataFrame(comparison_rows))
```

### P1. 연도별 보이스피싱 발생건수 변화

**연구 질문:** 2016~2025년 동안 보이스피싱 발생건수는 어떻게 변화했는가?

**사용 변수:** 연도, 기관사칭형_발생건수, 대출사기형_발생건수, 전체_발생건수

**분석 목적:** 보이스피싱이 지속적으로 발생하는 금융 피해인지 기본 추세를 확인한다.

```python
police_count_table = df_police_status[[
    '연도', '기관사칭형_발생건수', '대출사기형_발생건수', '전체_발생건수'
]].copy()
police_count_table['기관사칭형_비율(%)'] = (
    police_count_table['기관사칭형_발생건수'] / police_count_table['전체_발생건수'] * 100
).round(2)
police_count_table['대출사기형_비율(%)'] = (
    police_count_table['대출사기형_발생건수'] / police_count_table['전체_발생건수'] * 100
).round(2)

display(police_count_table)
```

```python
display(police_count_table[['연도', '전체_발생건수']])

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(police_count_table['연도'], police_count_table['전체_발생건수'], marker='o')
ax.set_title('연도별 전체 보이스피싱 발생건수')
ax.set_xlabel('연도')
ax.set_ylabel('발생건수(건)')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

### P3. 사기유형별 피해 규모 차이 - 발생건수

기관사칭형과 대출사기형의 연도별 발생건수를 같은 축에서 비교합니다. 이는 기본 현황 비교이며 통계적 유의성을 판단하지 않습니다.

### P2. 연도별 보이스피싱 피해금액 변화

**연구 질문:** 2016~2025년 동안 전체 피해금액은 어떻게 변화했는가?

**사용 변수:** 연도, 기관사칭형_피해액_억원, 대출사기형_피해액_억원, 전체_피해액_억원

**분석 목적:** 실제 금전적 피해 규모와 연도별 변화를 확인한다.

```python
police_amount_table = df_police_status[[
    '연도', '기관사칭형_피해액_억원', '대출사기형_피해액_억원', '전체_피해액_억원'
]].copy()
police_amount_table['기관사칭형_피해액_비율(%)'] = (
    police_amount_table['기관사칭형_피해액_억원'] / police_amount_table['전체_피해액_억원'] * 100
).round(2)
police_amount_table['대출사기형_피해액_비율(%)'] = (
    police_amount_table['대출사기형_피해액_억원'] / police_amount_table['전체_피해액_억원'] * 100
).round(2)

display(police_amount_table)
```

```python
display(police_amount_table[['연도', '전체_피해액_억원']])

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(police_amount_table['연도'], police_amount_table['전체_피해액_억원'], marker='o')
ax.set_title('연도별 전체 보이스피싱 피해금액')
ax.set_xlabel('연도')
ax.set_ylabel('피해금액(억원)')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

### P1·P2 종합. 연도별 발생건수 및 피해금액 복합그래프

P1과 P2에서 확인한 전체 발생건수와 전체 피해금액을 두 축으로 함께 비교합니다.

```python
police_combined_table = df_police_status.loc[
    df_police_status['연도'].between(2016, 2025),
    ['연도', '전체_발생건수', '전체_피해액_억원'],
].sort_values('연도').copy()

years = police_combined_table['연도'].astype(int)
total_counts = police_combined_table['전체_발생건수']
total_amounts = police_combined_table['전체_피해액_억원']
bar_colors = ['#4C78A8'] * len(police_combined_table)
bar_colors[years.tolist().index(2025)] = '#E45756'

fig, ax_amount = plt.subplots(figsize=(13, 7))
ax_count = ax_amount.twinx()

bars = ax_amount.bar(
    years, total_amounts, width=0.62, color=bar_colors, alpha=0.88,
    label='피해금액', zorder=2,
)
count_line, = ax_count.plot(
    years, total_counts, color='#222222', marker='o', markersize=6,
    linewidth=2.2, label='발생건수', zorder=3,
)

# 두 지표의 값 라벨이 같은 위치에 모이지 않도록 축별 여백과 오프셋을 분리합니다.
ax_amount.set_ylim(0, total_amounts.max() * 1.18)
ax_count.set_ylim(0, total_counts.max() * 1.16)
for bar, year, amount in zip(bars, years, total_amounts):
    is_2025 = year == 2025
    ax_amount.annotate(
        f'{amount:,.0f}',
        xy=(bar.get_x() + bar.get_width() / 2, bar.get_height()),
        xytext=(0, 5), textcoords='offset points',
        ha='center', va='bottom', fontsize=9,
        color='#C23B32' if is_2025 else '#355C7D',
        fontweight='bold' if is_2025 else 'normal',
    )

for year, count in zip(years, total_counts):
    ax_count.annotate(
        f'{count:,.0f}', xy=(year, count),
        xytext=(0, 9), textcoords='offset points',
        ha='center', va='bottom', fontsize=9, color='#222222',
        bbox={'boxstyle': 'round,pad=0.15', 'facecolor': 'white', 'alpha': 0.78, 'edgecolor': 'none'},
    )

ax_amount.set_title('2016~2025년 보이스피싱 발생건수 및 피해금액 변화', fontsize=15, pad=16)
ax_amount.set_xlabel('연도')
ax_amount.set_ylabel('피해금액(억원)', color='#355C7D')
ax_count.set_ylabel('발생건수(건)', color='#222222')
ax_amount.set_xticks(years)
ax_amount.yaxis.set_major_formatter(FuncFormatter(lambda value, _: f'{value:,.0f}'))
ax_count.yaxis.set_major_formatter(FuncFormatter(lambda value, _: f'{value:,.0f}'))
ax_amount.grid(axis='y', alpha=0.25, zorder=0)
ax_amount.legend([bars, count_line], ['피해금액', '발생건수'], loc='upper left')

plt.tight_layout()
plt.show()
```

### 제안서용 최근 3개년 피해현황

경찰청 보이스피싱 현황 공공데이터 기준

```python
recent_police_table = df_police_status.loc[
    df_police_status['연도'].between(2023, 2025),
    ['연도', '전체_발생건수', '전체_피해액_억원'],
].sort_values('연도').copy()

# 원본 DataFrame에서 계산된 최근 3개년 값이 제안서 검증값과 일치하는지 확인합니다.
expected_recent_values = {
    2023: {'전체_발생건수': 18902, '전체_피해액_억원': 4472},
    2024: {'전체_발생건수': 20839, '전체_피해액_억원': 8545},
    2025: {'전체_발생건수': 23360, '전체_피해액_억원': 12578},
}
actual_recent_values = (
    recent_police_table.assign(연도=recent_police_table['연도'].astype(int))
    .set_index('연도')[['전체_발생건수', '전체_피해액_억원']]
    .to_dict(orient='index')
)
if actual_recent_values != expected_recent_values:
    raise ValueError(
        '2023~2025년 값이 검증값과 다릅니다. '
        f'원본 DataFrame과 파생변수를 확인하세요: {actual_recent_values}'
    )

recent_years = recent_police_table['연도'].astype(int)
recent_counts = recent_police_table['전체_발생건수']
recent_amounts = recent_police_table['전체_피해액_억원']
recent_bar_colors = ['#4C78A8', '#4C78A8', '#E45756']

fig, ax_amount_recent = plt.subplots(figsize=(10, 6))
ax_count_recent = ax_amount_recent.twinx()

recent_bars = ax_amount_recent.bar(
    recent_years, recent_amounts, width=0.58,
    color=recent_bar_colors, alpha=0.9, label='피해금액', zorder=2,
)
recent_count_line, = ax_count_recent.plot(
    recent_years, recent_counts, color='#222222', marker='o', markersize=7,
    linewidth=2.4, label='발생건수', zorder=3,
)

# 두 축 모두 0부터 시작해 최근 3개년의 차이를 과장하지 않습니다.
ax_amount_recent.set_ylim(0, recent_amounts.max() * 1.18)
ax_count_recent.set_ylim(0, recent_counts.max() * 1.18)

for bar, year, amount in zip(recent_bars, recent_years, recent_amounts):
    is_2025 = year == 2025
    ax_amount_recent.annotate(
        f'{amount:,.0f}억원',
        xy=(bar.get_x() + bar.get_width() / 2, bar.get_height()),
        xytext=(0, -12), textcoords='offset points',
        ha='center', va='top', fontsize=11, color='white',
        fontweight='bold' if is_2025 else 'normal',
    )

for year, count in zip(recent_years, recent_counts):
    is_2025 = year == 2025
    ax_count_recent.annotate(
        f'{count:,.0f}건', xy=(year, count),
        xytext=(0, 10), textcoords='offset points',
        ha='center', va='bottom', fontsize=10, color='#222222',
        fontweight='bold' if is_2025 else 'normal',
    )

ax_amount_recent.set_title(
    '2023~2025년 보이스피싱 발생건수 및 피해금액 변화', fontsize=15, pad=16,
)
ax_amount_recent.set_xlabel('연도')
ax_amount_recent.set_ylabel('피해금액(억원)', color='#355C7D')
ax_count_recent.set_ylabel('발생건수(건)', color='#222222')
ax_amount_recent.set_xticks(recent_years)
ax_amount_recent.yaxis.set_major_formatter(FuncFormatter(lambda value, _: f'{value:,.0f}'))
ax_count_recent.yaxis.set_major_formatter(FuncFormatter(lambda value, _: f'{value:,.0f}'))
ax_amount_recent.grid(axis='y', alpha=0.25, zorder=0)
ax_amount_recent.legend(
    [recent_bars, recent_count_line], ['피해금액', '발생건수'], loc='upper left',
)

plt.tight_layout()
plt.show()
```

### 제안서용 연도별 건당 피해금액 추이

경찰청 공식 집계통계의 `전체_피해액_억원`과 `전체_발생건수`를 이용해 건당 피해금액을 자체 계산합니다.

```python
police_per_case_table = df_police_status.loc[
    df_police_status['연도'].between(2016, 2025),
    ['연도', '전체_발생건수', '전체_피해액_억원'],
].sort_values('연도').copy()

# 피해액의 억원 단위를 만원으로 바꾼 뒤 전체 발생건수로 나눕니다.
police_per_case_table['건당_피해금액_만원'] = (
    police_per_case_table['전체_피해액_억원'] * 10000
    / police_per_case_table['전체_발생건수']
)

display(police_per_case_table[['연도', '건당_피해금액_만원']].round(1))

years = police_per_case_table['연도'].astype(int)
per_case_amounts = police_per_case_table['건당_피해금액_만원']

fig, ax = plt.subplots(figsize=(13, 7))
ax.plot(
    years, per_case_amounts, color='#4C78A8', marker='o', markersize=7,
    linewidth=2.4, zorder=2,
)

# 최근 값은 같은 추세선 위에서 색과 크기만 달리해 과도하지 않게 강조합니다.
latest_value = per_case_amounts.loc[years == 2025].iloc[0]
ax.scatter(2025, latest_value, color='#E45756', s=95, zorder=3)

# 값이 비슷한 2021~2023년의 라벨을 위아래로 나눠 겹침을 피합니다.
label_offsets = {2021: (0, 12), 2022: (0, -20), 2023: (0, 12), 2025: (-8, 12)}
for year, value in zip(years, per_case_amounts):
    dx, dy = label_offsets.get(year, (0, 10))
    is_2025 = year == 2025
    ax.annotate(
        f'{value:,.0f}만원', xy=(year, value),
        xytext=(dx, dy), textcoords='offset points',
        ha='right' if is_2025 else 'center',
        va='top' if dy < 0 else 'bottom',
        fontsize=10,
        color='#C23B32' if is_2025 else '#355C7D',
        fontweight='bold' if is_2025 else 'normal',
        bbox={
            'boxstyle': 'round,pad=0.25', 'facecolor': 'white',
            'alpha': 0.9, 'edgecolor': '#E45756' if is_2025 else 'none',
        },
    )

ax.set_title('2016~2025년 보이스피싱 건당 피해금액 변화', fontsize=15, pad=16)
ax.set_xlabel('연도')
ax.set_ylabel('건당 피해금액(만원)')
ax.set_xticks(years)
ax.set_ylim(0, per_case_amounts.max() * 1.18)
ax.yaxis.set_major_formatter(FuncFormatter(lambda value, _: f'{value:,.0f}'))
ax.grid(axis='y', alpha=0.25)

plt.tight_layout()
plt.show()
```

**해석**

- 건당 피해금액은 경찰청 공식 집계통계의 `전체 피해금액 ÷ 전체 발생건수`로 자체 계산한 값이며, 경찰청이 별도 KPI로 직접 발표한 수치가 아닙니다.
- 2016년 약 862만원/건에서 2025년 약 5,384만원/건으로 커졌습니다.
- 장기적으로 변동이 있었으며, 최근 2024~2025년에 건당 금전적 피해규모가 크게 나타났습니다. 발생건수뿐 아니라 건당 피해금액도 중요하게 볼 필요가 있습니다.

### P8. 지역별 건당 피해금액 비교

**연구 질문**

- 지역별로 사건 1건당 피해금액 규모가 어떻게 다른가?

**분석 목적**

- 단순 총 발생건수나 총 피해금액만으로는 알기 어려운 사건당 피해규모를 확인한다.

**계산 방법**

- 연도별 건당 피해금액: `피해금액(억원) / 발생건수 × 10,000`으로 계산하여 만원/건 단위로 표시한다.
- 2023~2025년 전체 값은 연도별 값의 단순 평균이 아니라 `3개년 총 피해금액 / 3개년 총 발생건수`로 계산한다.

**주의사항**

- 지역별 인구 규모를 보정한 피해 위험률이 아니다.
- 지역 자체가 피해금액의 원인이라는 의미가 아니다.
- 표본 규모가 서로 다르므로 작은 지역의 건당 피해액을 과도하게 해석하지 않는다.
- 피해금액 원본 CSV에는 단위가 직접 명시되어 있지 않다. 지역 합계와 전체 피해액(억원)의 차이는 2023년 1, 2024년 1, 2025년 0으로 거의 일치하므로 분석상 억원 단위로 해석하되, 이를 원출처 메타데이터에서 공식 확인한 것으로 단정하지 않는다.

```python
p8_year_columns = ['2023년', '2024년', '2025년']

# 지역별 피해금액의 연도 합계와 억원 단위 전체 피해액을 교차확인합니다.
p8_unit_check_rows = []
for year_column in p8_year_columns:
    year = int(year_column.replace('년', ''))
    region_amount_sum = df_police_region_amount[year_column].sum()
    total_amount_row = df_police_status.loc[df_police_status['연도'] == year, '전체_피해액_억원']
    if len(total_amount_row) != 1:
        raise ValueError(f'{year}년 전체 피해액 행을 하나로 확인할 수 없습니다.')
    total_amount = total_amount_row.iloc[0]
    p8_unit_check_rows.append({
        '연도': year,
        '지역별 피해금액 합계': region_amount_sum,
        '전체 피해액(억원)': total_amount,
        '차이(지역 합계-전체)': region_amount_sum - total_amount,
    })

p8_unit_check_table = pd.DataFrame(p8_unit_check_rows)
p8_unit_consistent = p8_unit_check_table['차이(지역 합계-전체)'].abs().le(1).all()

display(p8_unit_check_table)
print('연도별 합계 차이가 모두 1 이하인가:', p8_unit_consistent)
if not p8_unit_consistent:
    raise ValueError('지역별 피해금액과 전체 피해액의 합계 차이가 예상 범위를 벗어났습니다. 단위를 재검토하세요.')
print('※ 전체 피해액(억원)과 연도별 합계가 거의 일치하여 분석상 억원 단위로 해석합니다.')
print('※ 다만 지역별 피해금액 원본 컬럼에는 단위가 직접 명시되어 있지 않습니다.')
```

```python
p8_region_count_source = df_police_region_count[['시도청'] + p8_year_columns].copy()
p8_region_amount_source = df_police_region_amount[['시도청'] + p8_year_columns].copy()

count_only_regions = sorted(set(p8_region_count_source['시도청']) - set(p8_region_amount_source['시도청']))
amount_only_regions = sorted(set(p8_region_amount_source['시도청']) - set(p8_region_count_source['시도청']))
print('발생건수에만 있는 지역:', count_only_regions)
print('피해금액에만 있는 지역:', amount_only_regions)

# outer 병합으로 어느 한쪽에만 존재하는 지역도 놓치지 않고 검증합니다.
p8_region_merged_table = p8_region_count_source.merge(
    p8_region_amount_source,
    on='시도청',
    how='outer',
    suffixes=('_발생건수', '_피해금액'),
    validate='one_to_one',
    indicator=True,
)
if not p8_region_merged_table['_merge'].eq('both').all():
    missing_regions = p8_region_merged_table.loc[p8_region_merged_table['_merge'] != 'both', ['시도청', '_merge']]
    raise ValueError(f'발생건수와 피해금액이 매칭되지 않는 지역이 있습니다.\n{missing_regions}')
p8_region_merged_table = p8_region_merged_table.drop(columns='_merge')

# 결과표는 원본 단위를 보존하고, 건당 값만 사람이 읽기 쉬운 만원/건으로 환산합니다.
p8_region_summary_table = pd.DataFrame({'시도청': p8_region_merged_table['시도청']})
for year_column in p8_year_columns:
    count_column = f'{year_column}_발생건수'
    amount_column = f'{year_column}_피해금액'
    year_count_values = p8_region_merged_table[count_column]
    safe_year_count_values = year_count_values.where(year_count_values.ne(0))
    p8_region_summary_table[f'{year_column} 발생건수'] = year_count_values
    p8_region_summary_table[f'{year_column} 피해금액'] = p8_region_merged_table[amount_column]
    p8_region_summary_table[f'{year_column} 건당 피해금액'] = (
        p8_region_merged_table[amount_column] / safe_year_count_values * 10_000
    )

count_result_columns = [f'{year_column} 발생건수' for year_column in p8_year_columns]
amount_result_columns = [f'{year_column} 피해금액' for year_column in p8_year_columns]
p8_region_summary_table['2023~2025 총 발생건수'] = p8_region_summary_table[count_result_columns].sum(axis=1)
p8_region_summary_table['2023~2025 총 피해금액'] = p8_region_summary_table[amount_result_columns].sum(axis=1)
safe_total_count_values = p8_region_summary_table['2023~2025 총 발생건수'].where(
    p8_region_summary_table['2023~2025 총 발생건수'].ne(0)
)
p8_region_summary_table['2023~2025 건당 피해금액'] = (
    p8_region_summary_table['2023~2025 총 피해금액'] / safe_total_count_values * 10_000
)

per_case_columns = [column for column in p8_region_summary_table.columns if '건당 피해금액' in column]
p8_region_summary_table[per_case_columns] = p8_region_summary_table[per_case_columns].round(2)
p8_region_summary_table = p8_region_summary_table.sort_values(
    '2023~2025 건당 피해금액', ascending=False, na_position='last'
).reset_index(drop=True)

zero_count_rows = p8_region_summary_table.loc[
    p8_region_summary_table[count_result_columns].eq(0).any(axis=1),
    ['시도청'] + count_result_columns,
]
print('발생건수 0이 있는 지역-연도 수:', len(zero_count_rows))
print('[단위] 발생건수: 건, 피해금액: 억원(합계 일관성에 따른 분석상 해석), 건당 피해금액: 만원/건')
display(p8_region_summary_table)
```

```python
p8_region_plot_table = p8_region_summary_table.sort_values('2023~2025 건당 피해금액')

fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(p8_region_plot_table['시도청'], p8_region_plot_table['2023~2025 건당 피해금액'])
ax.set_title('2023~2025년 지역별 건당 피해금액')
ax.set_xlabel('건당 피해금액(만원/건)')
ax.set_ylabel('시도청')
ax.xaxis.set_major_formatter(FuncFormatter(lambda value, position: f'{value:,.0f}'))
plt.tight_layout()
plt.show()
```

- 기관사칭형 피해금액 비중은 2024년 62.60%에서 2025년 78.58%로 커졌습니다.
- 기관사칭형 건당 피해금액은 2024년 5,619.29만원에서 2025년 7,418.75만원으로 증가했습니다. 같은 기간 대출사기형은 2,823.32만원에서 2,684.07만원으로 감소했습니다.

따라서 **전체 기간의 구조와 최근의 구조는 다를 수 있습니다.** 전체 누적에서는 대출사기형의 규모가 더 크지만, 최근에는 기관사칭형의 발생 비중과 피해 비중이 커지고 건당 피해규모도 대출사기형보다 크게 나타났습니다. 두 관점을 하나의 결론으로 섞지 않고 구분하여 해석해야 합니다.

## 분석 해석 시 주의사항

- 경찰청 데이터는 집계통계입니다.
- 개별 피해자의 위험을 예측하는 데이터가 아닙니다.
- 연령별 자료는 전체 인구 대비 피해율이 아닙니다. 따라서 특정 연령대가 보이스피싱에 걸릴 확률이 높다고 표현하지 않습니다.
- 연령 자료는 전체 피해자 중 각 연령대가 차지하는 구성으로 해석합니다.
- 지역별 발생건수는 지역 인구가 보정되지 않은 총 발생건수입니다.
- 지역별 피해금액도 지역 위험확률을 의미하지 않습니다.
- P8 건당 피해금액도 지역 자체가 피해규모 차이의 원인이라는 의미가 아닙니다.
- 데이터에서 나타난 통계적 추세를 인과관계로 표현하지 않습니다.
- 현재 분석은 미래 발생건수나 피해금액을 예측하는 분석이 아닙니다.
- 지역별 피해금액 자료는 원본 컬럼에 단위가 직접 명시되지 않았습니다. 다만 전체 피해액 자료와 연도별 합계가 거의 일치하는 것을 근거로 분석상 억원 단위로 해석했습니다.

## 경찰청 데이터 분석 완료

- 0단계 데이터 준비 완료
- 1단계 기본 EDA(P1~P8) 완료
- 경찰청 데이터 분석은 현재 범위에서 종료
- 경찰청 데이터는 보이스피싱 현황·추세 설명용으로 활용
- 2단계 이후 추가 통계검정은 진행하지 않음
- 경찰청 데이터 기반 ML은 진행하지 않음
- 이후 대시보드용 결과 선별 및 시각화 작업으로 이동

경찰청 데이터는 개별 피해사례가 아닌 집계통계이므로 현재 프로젝트에서는 추가 관계 검정이나 ML 모델로 확장하지 않고, 대한민국 보이스피싱의 규모와 변화 추세를 설명하는 근거 데이터로 활용합니다.
